# Net liquidity Calculator 
The following code calculates the net liquidity using data from **FRED reserve** and uses api key , to see code running add you fred api key


```python
!pip install fredapi
from fredapi import Fred
import pandas as pd

api_key = "Enter your Key"

if not api_key:
    raise ValueError("Missing FRED_API_KEY! Set it in your environment.")

fred = Fred(api_key=api_key)
```

```python
class FedLiquidityMonitor:
    def __init__(self, fred):
        self.fred = fred

    def fetch_data(self, raw_series):
        try:
            series = self.fred.get_series(raw_series)
        except Exception as e:
            print(f"Technical issue try again later {e}")
            return None
        return series

    def get_metadata(self, raw_series):
        try:
            metadata = self.fred.get_series_info(raw_series)
        except Exception as e:
            print(f"Technical issue try again later {e}")
            return None
        return metadata

    def convert_to_billions(self, raw_series):
        series = self.fetch_data(raw_series)
        metadata = self.get_metadata(raw_series)

        if "Billions" in metadata["units"]:
            upd_series = series
        elif "Millions" in metadata["units"]:
            upd_series = series / 1000
        elif "Thousands" in metadata["units"]:
            upd_series = series / 1000000
        else:
            print("Try Again Later!!!")
            return None

        return upd_series

    def prepare_dataframe(self, raw_series):
        series = self.convert_to_billions(raw_series)
        data = pd.concat([series], axis=1)
        data = data.ffill()
        data.columns = [raw_series]
        return data
```
